```python
from wordcloud import WordCloud, STOPWORDS, ImageColorGenerator
import numpy as np
import matplotlib.pyplot as plt
import PIL.Image

import pandas as pd
```

```python
monkey_pox = pd.read_csv('datasets/reddit_monkeypox.csv')
```

```python
text = ' '.join(text for text in monkey_pox['title'])

custom_stopwords = set(STOPWORDS)
custom_stopwords.update(
    [
        "reddit",
        "post",
        "comment",
        "deleted",
        "removed",
        "https",
        "com",
        "amp",
        "people",
        "one",
    ]
)
```

```python
wordcloud = WordCloud(
    width=1000,
    height=500,
    background_color="white",  # Dark mode canvas looks great with social data
    max_words=150,  # Limits clutter
    stopwords=custom_stopwords,
    colormap="plasma",  # Vivid color profile for contrast
    random_state=42,  # Ensures the layout remains identical every run
).generate(text)
```

```python
plt.figure(figsize=(15, 7.5))
plt.imshow(wordcloud, interpolation="bilinear")
plt.axis("off")  # Hides graph axis grids
plt.tight_layout(pad=0)

# Display the output window
plt.show()
```
