```python
import pandas as pd
import plotly.express as px

# 1. Create a dummy project dataset
data = [
    {
        "Task": "Project Planning",
        "Start": "2026-03-01",
        "End": "2026-03-07",
        "Phase": "Design",
        "Completion": 100,
    },
    {
        "Task": "Architecture Layout",
        "Start": "2026-03-06",
        "End": "2026-03-15",
        "Phase": "Design",
        "Completion": 85,
    },
    {
        "Task": "Core Engine Coding",
        "Start": "2026-03-12",
        "End": "2026-04-05",
        "Phase": "Development",
        "Completion": 40,
    },
    {
        "Task": "API Integration",
        "Start": "2026-03-20",
        "End": "2026-04-12",
        "Phase": "Development",
        "Completion": 10,
    },
    {
        "Task": "System Testing",
        "Start": "2026-04-10",
        "End": "2026-04-25",
        "Phase": "QA / DevOps",
        "Completion": 0,
    },
    {
        "Task": "Deployment",
        "Start": "2026-04-23",
        "End": "2026-04-30",
        "Phase": "QA / DevOps",
        "Completion": 0,
    },
]

df = pd.DataFrame(data)

# Convert string dates to datetime objects (Required for Timeline)
df["Start"] = pd.to_datetime(df["Start"])
df["End"] = pd.to_datetime(df["End"])

# 2. Define custom colors for project phases
phase_colors = {
    "Design": "#3A86FF",  # Vibrant Blue
    "Development": "#FF006E",  # Vivid Pink
    "QA / DevOps": "#8338EC",  # Deep Purple
}

# 3. Build the Gantt timeline chart
fig = px.timeline(
    df,
    x_start="Start",
    x_end="End",
    y="Task",
    color="Phase",
    color_discrete_map=phase_colors,
    title="🏢 Custom Corporate Project Roadmap",
    hover_data={
        "Completion": ":.0f%",
        "Start": "|%B %d, %Y",
        "End": "|%B %d, %Y",
    },  # Clean date format in tooltip
)

# 4. Apply Advanced Layout Customizations
fig.update_layout(
    xaxis_title="Timeline",
    yaxis_title=None,  # Hiding redundant label
    yaxis={"categoryorder": "array", "categoryarray": df["Task"][::-1]},
    # ^ Forces chart to read chronologically from top-to-bottom
    plot_bgcolor="white",  # Clean backdrop
    hovermode="closest",
    font=dict(family="Arial, sans-serif", size=13),
    legend=dict(
        title="Project Phases",
        orientation="h",
        yanchor="bottom",
        y=1.02,
        xanchor="right",
        x=1,
    ),
)

# 5. Add custom grid lines and alternating background clean aesthetics
fig.update_xaxes(
    showgrid=True,
    gridcolor="#EBF0F5",
    tickformat="%b %d\n%Y",  # Multi-line date ticks
)

fig.update_yaxes(showgrid=False)

# Launch chart in browser/notebook
fig.show()
```
